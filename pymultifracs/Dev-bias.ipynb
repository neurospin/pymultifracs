```python
from pymultifracs import wavelet_analysis, mfa
from pymultifracs.simul import mrw, fbm
import matplotlib.pyplot as plt
import numpy as np
```

```python
X = mrw((2**15, 100), .8, .22)
WTpL = wavelet_analysis(X).get_leaders(2)
```

```python
X = fbm((2**15, 100), .8)
WTpL = wavelet_analysis(X).get_leaders(2)
```

```python
pwt = mfa(WTpL, [(6, 11)], bias_correction=False)
```

```python
pwt_corr = mfa(WTpL, [(6, 11)], bias_correction=True)
```

```python
pwt_corr.cumulants.c2.mean()
```

```python
pwt.cumulants.c2.mean()
```

```python
pwt.cumulants.plot()
```

```python
pwt_corr.cumulants.plot()
```
